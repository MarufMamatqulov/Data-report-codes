```python
import pandas as pd
```

```python
df = pd.read_excel("Таблица_компенсированных_сертификатов_3.xlsx")
df
```

```python
df["Сертификат тури"].value_counts()
```

```python
def calculate_salary_statistics(df):
    # Remove outliers using IQR method
    Q1 = df['salary_2024'].quantile(0.25)
    Q3 = df['salary_2024'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Create clean dataframe without outliers
    df_clean = df[
        (df['salary_2024'] >= lower_bound) & 
        (df['salary_2024'] <= upper_bound)
    ]
    
    # Calculate statistics by specialization
    salary_stats = df_clean.groupby('Сертификат тури').agg({
        'salary_2024': [
            ('median', 'median'),
            ('count', 'count'),
            ('Q1', lambda x: x.quantile(0.25)),
            ('Q3', lambda x: x.quantile(0.75))
        ]
    }).round(2)
    
    # Sort by median salary
    salary_stats = salary_stats.sort_values(('salary_2024', 'median'), ascending=False)
    
    return salary_stats

# Example usage 
stats = calculate_salary_statistics(df)
stats
```

```python
stats.to_excel("stats.xlsx")
```
