```python
df_new = pd.read_excel("crm.xlsx", sheet_name="new")
df_old = pd.read_excel("crm.xlsx", sheet_name="old")
```

```python
inn_new = set(df_new['INN'])
inn_old = set(df_old['INN'])

removed_inn = inn_old - inn_new
df_removed = df_old[df_old['INN'].isin(removed_inn)]

added_inn = inn_new - inn_old
df_added = df_new[df_new['INN'].isin(added_inn)]

df_removed = df_removed.sort_values('INN').reset_index(drop=True)
df_added = df_added.sort_values('INN').reset_index(drop=True)

print(f"Number of removed records: {len(df_removed)}")
print(f"Number of added records: {len(df_added)}")
```

```python
# Export to Excel if needed
df_removed.to_excel('removed_records.xlsx', index=False)
df_added.to_excel('added_records.xlsx', index=False)
```
