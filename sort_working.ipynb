```python
df1 = pd.read_excel("Content-ishga-joylashgan_joylashmagan.xlsx")
df1
```

```python
import pandas as pd


# Find rows in `df` that are not already in `df1` based on the key columns
new_data = df1.merge(df[['Fish', 'Sertifikat']], on=['Fish', 'Sertifikat'], how='left', indicator=True)

# Filter rows that are unique to `df`
new_data = new_data[new_data['_merge'] == 'left_only']

# Drop the indicator column
new_data.drop(columns=['_merge'], inplace=True)

# Create a separate DataFrame to track the added rows
added_rows = new_data.copy()

# Append unique rows to `df1`
df1_updated = pd.concat([df1, new_data], ignore_index=True)

# Drop duplicates from the final DataFrame
df1_updated = df1_updated.drop_duplicates(subset=['Fish', 'Sertifikat'], keep='first')

# Display the DataFrame containing the added rows
print("Added Rows:")
added_rows


```

```python
# Display the updated `df1`
print("Updated df1:")
df1_updated
```
