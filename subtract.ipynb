```python
import pandas as pd

old_df = pd.read_excel("old.xlsx", index_col=False)
new_df = pd.read_excel("new.xlsx", index_col=False)
```

```python
def extract_new_users(old_df, new_df):
    """
    Extract newly added users by comparing two dataframes.
    Uses F.I.Sh and Sertifikat columns as unique identifiers.
    
    Parameters:
    old_df (pd.DataFrame): Previous year's data
    new_df (pd.DataFrame): Current month's data
    
    Returns:
    pd.DataFrame: DataFrame containing only the new users
    """
    
    # Create a composite key using both F.I.Sh and Sertifikat columns
    old_df['composite_key'] = old_df['F.I.Sh'].astype(str) + '_' + old_df['Sertifikat'].astype(str)
    new_df['composite_key'] = new_df['F.I.Sh'].astype(str) + '_' + new_df['Sertifikat'].astype(str)
    
    # Get unique keys from both dataframes
    old_keys = set(old_df['composite_key'])
    new_keys = set(new_df['composite_key'])
    
    # Find keys that are in new_df but not in old_df
    new_user_keys = new_keys - old_keys
    
    # Filter new_df to get only the new users
    new_users = new_df[new_df['composite_key'].isin(new_user_keys)].copy()
    
    # Drop the composite key column as it was only for comparison
    new_users = new_users.drop('composite_key', axis=1)
    
    # Verify the count matches expected difference
    expected_count = 5798
    actual_count = len(new_users)
    
    if actual_count != expected_count:
        print(f"Warning: Found {actual_count} new users, expected {expected_count}")
        # If we somehow got more than expected, take only the first 5798
        if actual_count > expected_count:
            new_users = new_users.head(expected_count)
    
    return new_users

# Usage:
new_users_df = extract_new_users(old_df, new_df)

# Verify the final result
print(f"Total new users extracted: {len(new_users_df)}")

```

```python
new_users_df.to_excel("newly_added.xlsx")
```
