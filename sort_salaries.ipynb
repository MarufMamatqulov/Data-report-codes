```python
import pandas as pd

def load_salary_data(file_path):
    """Load and prepare salary data"""
    return pd.read_excel(file_path)['salary_2024']

def calculate_bounds(salaries, lower_percentile=0.05, upper_percentile=0.95):
    """Calculate percentile bounds"""
    return (salaries.quantile(lower_percentile), 
            salaries.quantile(upper_percentile))

def format_stats(data_series, suffix='M'):
    """Format statistics in millions with proper formatting"""
    stats = {
        'Min': data_series.min() / 1_000_000,
        'Max': data_series.max() / 1_000_000,
        'Mean': data_series.mean() / 1_000_000,
        'Median': data_series.median() / 1_000_000,
        'Standard Deviation': data_series.std() / 1_000_000
    }
    return {k: f"{v:.2f}{suffix}" for k, v in stats.items()}

file_path = 'Misol.xlsx'
salaries = load_salary_data(file_path)
lower_bound, upper_bound = calculate_bounds(salaries)

filtered_salaries = salaries[(salaries >= lower_bound) & 
                           (salaries <= upper_bound)]

original_stats = format_stats(salaries)
filtered_stats = format_stats(filtered_salaries)

print("\nOriginal Statistics:")
for key, value in original_stats.items():
    print(f"{key}: {value}")

print("\nFiltered Statistics:")
for key, value in filtered_stats.items():
    print(f"{key}: {value}")
```
