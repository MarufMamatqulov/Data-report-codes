```python
import pandas as pd

# Kiril-Lotin o'girish funksiyasi
def kiril_to_lotin(text):
    if not isinstance(text, str):
        return text

    translit_map = {
        'А': 'A','Қ':'Q','Ғ':'G`', 'Б': 'B', 'В': 'V', 'Г': 'G', 'Д': 'D', 'Е': 'E', 'Ё': 'Yo', 'Ж': 'J', 'З': 'Z', 'И': 'I',
        'Й': 'Y', 'К': 'K', 'Л': 'L', 'М': 'M', 'Н': 'N', 'О': 'O', 'П': 'P', 'Р': 'R', 'С': 'S', 'Т': 'T',
        'У': 'U', 'Ф': 'F', 'Х': 'X', 'Ц': 'Ts', 'Ч': 'Ch', 'Ш': 'Sh', 'Щ': 'Sh', 'Ъ': '', 'Ы': 'I', 'Ь': '',
        'Э': 'E', 'Ю': 'Yu', 'Я': 'Ya',
        'а': 'a', 'қ':'q','ғ':'g`', 'б': 'b', 'в': 'v', 'г': 'g', 'д': 'd', 'е': 'e', 'ё': 'yo', 'ж': 'j', 'з': 'z', 'и': 'i',
        'й': 'y', 'к': 'k', 'л': 'l', 'м': 'm', 'н': 'n', 'о': 'o', 'п': 'p', 'р': 'r', 'с': 's', 'т': 't',
        'у': 'u', 'ф': 'f', 'х': 'x', 'ц': 'ts', 'ч': 'ch', 'ш': 'sh', 'щ': 'sh', 'ъ': '', 'ы': 'i', 'ь': '',
        'э': 'e', 'ю': 'yu', 'я': 'ya'
    }

    result = ''.join(translit_map.get(char, char) for char in text)
    return result

# Excel faylni yuklash
file_path = 'xalqaro IT sertifikat (2).xlsx'  # bu faylni o'zingizning faylingizga mos ravishda o'zgartiring
df = pd.read_excel(file_path)

# Har bir katakchani transliteratsiya qilish
df_latin = df.applymap(kiril_to_lotin)

# Natijani yangi faylga saqlash
new_file_path = file_path.replace('.xlsx', '_lotin.xlsx')
df_latin.to_excel(new_file_path, index=False)

print(f"✅ Translatsiya tugadi! Yangi fayl saqlandi: {new_file_path}")
```
