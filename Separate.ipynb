```python
import pandas as pd
```

```python
df = pd.read_csv('lang_level1.csv')
df
```

```python
from datetime import datetime

df['b_date'] = pd.to_datetime(df['b_date'], format='%d.%m.%Y')

# Calculate age in years as of today
today = datetime.today()
df['age'] = (today - df['b_date']).dt.days / 365.25

# Filter the dataframe to keep only people 30 years old or younger
df_filtered = df[df['age'] <= 30]

# Drop the age column if you don't need it anymore
df_filtered = df_filtered.drop(columns=['age'])

# Print the filtered DataFrame
df_filtered
```

```python

# Create the 'degree' column based on 'cert_common_ball'
df_filtered['degree'] = pd.cut(df_filtered['cert_common_ball'],
                        bins=[0, 38, 51, 65, float('inf')], 
                        labels=['Lower than B1', 'B1', 'B2', 'C1 va undan yuqori'],
                        include_lowest=True)

df_filtered
```

```python
df_b2 = df_filtered[~((df_filtered['cert_subject_name'] == 'Ingliz tili') & (df_filtered['cert_common_ball'] < 55) | (df_filtered['cert_common_ball'] > 64))].copy()
df_b2['b_date'] = df_b2['b_date'].dt.strftime('%d.%m.%Y')
df_b2 = df_b2[df_b2['cert_subject_name'] == 'Ingliz tili']
df_b2.to_excel('lang_level_55-64.xlsx', index=False)
```

```python
import pandas as pd

# Assuming your DataFrame is named 'df'

# Filter out English certificates with scores below 55
new_df = df_filtered[~((df_filtered['cert_subject_name'] == 'Ingliz tili') & (df_filtered['cert_common_ball'] <= 55))].copy()
new_df['b_date'] = new_df['b_date'].dt.strftime('%d.%m.%Y') 
new_df
```

```python
df_c1_above_except_english = new_df[((new_df['degree'] == 'C1 va undan yuqori') & (new_df['cert_subject_name'] != 'Ingliz tili')) | (new_df['cert_subject_name'] == 'Ingliz tili')]
df_c1_above_except_english.to_excel('c1_va_undan_yuqori.xlsx', index=False)
```
