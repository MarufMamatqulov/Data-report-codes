```python
import pandas as pd
import re

def normalize_course_name(course_name):
   # Convert to string and lowercase
   course_name = str(course_name).lower()
   
   # Remove extra spaces and special characters
   course_name = re.sub(r'\s+', ' ', course_name)
   course_name = re.sub(r'[^\w\s]', '', course_name)
   
   # Handle multiple courses
   if ',' in course_name:
       course_name = course_name.split(',')[0]
   
   # Comprehensive normalization map
   normalization_map = {
       'frontend': 'front-end',
       'frontendweb': 'front-end',
       'front end': 'front-end',
       'front end web': 'front-end',
       
       'fullstack': 'full stack',
       'full stack web': 'full stack',
       'full stack development': 'full stack',
       
       'data science': 'data science',
       'data science va ai': 'data science',
       'datascience': 'data science',
       
       'backend': 'back-end',
       'backendweb': 'back-end',
       'back end': 'back-end',
       
       'qa': 'quality assurance',
       'quality assurance testing': 'quality assurance',
       
       'uxui': 'ux/ui',
       'ui ux': 'ux/ui',
       
       'graphic': 'design',
       'graphic design': 'design',
       
       'kali linux': 'linux'
   }
   
   # Apply normalization
   for key, value in normalization_map.items():
       if key in course_name:
           course_name = value
           break
   
   return course_name.strip()

# Read Excel file
df = pd.read_excel('graduates.xlsx')

# Normalize courses
df['course'] = df['Normalized_Course'].apply(normalize_course_name)

# Group and count unique normalized courses
course_distribution = df['course'].value_counts().reset_index()
course_distribution.columns = ['Course', 'Count']

# Save to Excel
course_distribution.to_excel('course_distribution.xlsx', index=False)

# Display results
print(course_distribution)
```
